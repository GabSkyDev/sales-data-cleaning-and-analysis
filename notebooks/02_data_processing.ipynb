# 3. Engenharia de atributos e Extração de Insights

```python
# Extração da base de dados limpa com módulo 'data_cleaning.py'
import sys
import os

sys.path.append(os.path.abspath(".."))

import pandas as pd
import matplotlib.pyplot as plt
from src.data_cleaning import data_cleaning

df = pd.read_csv("../data/raw_data.csv")
df = data_cleaning(df)
df.head()
```

```python
# Feature Engineering: Criação de coluna 'Total_Venda'
df['Total_Venda'] = df['Quantidade'] * df['Preco_Unitario']

df.head()
```

```python
# Data Insight: Qual o total da receita?
total_venda = df['Total_Venda'].sum()
print(f"Receita total de vendas da loja: R${total_venda:,.2f}")
```

```python
# Data Insight: Qual a receita total por categoria de produto?
receita_por_categoria = df.groupby('Categoria')['Total_Venda'].sum()
print("Receita total por categoria:")
receita_por_categoria
```

```python
# Data Insight: Quais os produtos mais vendidos em quantidade?
produto_mais_vendido = df.groupby('Produto')['Quantidade'].sum().sort_values(ascending=False)
print("Total de unidades vendidas por quantidade: ")
produto_mais_vendido
```

```python
# Data Insight: quantidade de vendas realizadas por dia?
df_temporal = df.set_index('Data_Compra')

vendas_por_dia = df_temporal.resample('D')['Total_Venda'].sum()
print("Vendas realizadas por dia:")
vendas_por_dia
```
